Problem 5
Write code to display only specific items from a list/array that satisfy the following conditions:
• The number must be divisible by five.
• If the number is greater than or equal to 150, then skip it.
• If the number is greater than 500, then stop the loop.
Example input:
• Python: [12,75,150,182,145,525,50

```python
numbers = [12, 75, 150, 182, 145, 525, 50]

for x in numbers:
    if x > 500:
        break
    if x >= 150:
        continue
    if x % 5 == 0:
        print(x)
```

Problem 7
Write code to print all prime numbers between any given pair of numbers n and m to the console.
Use as example input n = 25 and m = 50.
NOTE:A prime number is a natural number greater than 1 that can only be divided by itself and
1 without remainders. Examples are: 2, 3, 5, 7, 11.

```python
n = 25
m = 50

for num in range(n, m + 1):
    if num <= 1:
        continue
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```
