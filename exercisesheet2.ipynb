### Problem 10
Write code to print the following pattern for a maximum depth $n$ to the console.

``` {.bash language="bash"}
 *
 * *
 * * *
 * *
 *
```

````python
def pattern(n):
    i=1
    while i<=n:
        j=0
        while j<i:
            print("*", end="")
            j=j+1
        print()
        i=i+1
    i=n-1
    while i>=1:
        j=0
        while j<i:
            print("*", end="")
            j=j+1
        print()
        i=i-1

pattern(5)
````

### Problem 12
Write code to calculate the factorial of a given number $n$.

**NOTE:** The factorial of $n$ is denoted as $n!$ and is
calculated as $n \times n-1 \times n-2 \times \cdots \times  2 \times 1$.

````python
def nfact(n):
    x=1
    r=n
    while x<n:
        r=r*(n-x)
        x=x+1
    return r

print(nfact(5))
````

````python
def nfact(n):
   r=1
   for x in range(1,n+1):
         r=r*x
   return r

print(nfact(5)) 
````

### Problem 13
Write code to display the entire Fibbonacci sequence up to a given number $n$.

**NOTE:** The Fibbonacci sequence $F$ is a sequence of numbers where the current Fibbonacci number $F_i$ is the sum of the two preceeding Fibbonaci numbers $F_{i-2}+F_{i-1}$

````python
def fibonacci(n):
    fa=0
    fb=1
    
    while fa<=n:
       print(fa, end=" ")   
       fa, fb = fb, fa + fb

fibonacci(50)
````

### Problem 14
Write code that builds the complement sequence for a given DNA sequence. Print the result as follows:

``` {.bash language="bash"}
5'-ATCGTGAATCG-3'
    |||||||||||
 3'-TAGCACTTAGC-5'
```

````python
def complement(g):
    if g =="A":
        return "T"
    elif g =="T":
        return "A"
    elif g =="C":
        return "G"
    elif g =="G":
        return "C"
    else:
        return g

seq="ATCGTGAATCG"
output=""
for base in seq:
    output=output+complement(base)

print("5´-"+seq+"-3´")
print("   "+ "|" * len(seq))
print("5´-"+output+"-3´")


````

````python
def lls(nums: list[float]) -> tuple[float, int, int]:
    
    best_start=0
    best_end=0
    best_sum=nums[best_start]

    for start in range(0, len(nums)):
        current_sum=nums[start]
        for end in range(start+1, len(nums)):
            current_sum=current_sum+nums[end]
            if current_sum>best_sum:
                best_sum=current_sum
                best_start=start
                best_end=end
    return best_sum, best_start, best_end

print(lls([1, -2, 3, 4, -1, 2, 1, -5, 4])) 


print("Test 1: ", lls([1, -2, 3, 4, -1, 2, 1, -5, 4])) 


print("Test 2: ", lls([-1, -2, -3, -4]))


 
````

````python
def gcd(m, n, depth=0):
    if n > m:
        print("m > n branch at depth", depth)
        return gcd(n, m, depth + 1)
    elif m% n == 0:
        print("m % n == 0 branch at depth", depth)
        return n
    else:
        print("m % n != 0 branch at depth", depth)
        return gcd(n, m % n, depth + 1)

print(gcd(18, 48))
````
